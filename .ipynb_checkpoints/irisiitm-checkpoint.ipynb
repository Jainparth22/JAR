```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

# Load the dataset
X = pd.read_csv('irisX.csv', header=None).values  # Features
y = pd.read_csv('irisY.csv', header=None).values.ravel()  # Labels

# Use only the first two features
X = X[:, :2]

# Split into training (first 100 points) and test sets (remaining 50 points)
X_train, X_test = X[:100], X[100:]
y_train, y_test = y[:100], y[100:]

# Standardize the features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Initialize variables to store best accuracies
best_accuracy_l1 = 0
best_accuracy_l2 = 0

# Train Perceptron with L1 penalty
for alpha in [0.0001, 0.001, 0.01, 0.1, 1]:  # Regularization strengths
    perceptron_l1 = Perceptron(penalty='l1', alpha=alpha, max_iter=1000, random_state=42)
    perceptron_l1.fit(X_train, y_train)
    y_pred_l1 = perceptron_l1.predict(X_test)
    accuracy_l1 = accuracy_score(y_test, y_pred_l1)
    best_accuracy_l1 = max(best_accuracy_l1, accuracy_l1)

# Train Perceptron with L2 penalty
for alpha in [0.0001, 0.001, 0.01, 0.1, 1]:  # Regularization strengths
    perceptron_l2 = Perceptron(penalty='l2', alpha=alpha, max_iter=1000, random_state=42)
    perceptron_l2.fit(X_train, y_train)
    y_pred_l2 = perceptron_l2.predict(X_test)
    accuracy_l2 = accuracy_score(y_test, y_pred_l2)
    best_accuracy_l2 = max(best_accuracy_l2, accuracy_l2)

# Print results rounded to two decimal places
print(f"Best accuracy with L1 penalty: {round(best_accuracy_l1, 2)}")
print(f"Best accuracy with L2 penalty: {round(best_accuracy_l2, 2)}")
```

```python
# Import required libraries
import pandas as pd
import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score

# Read the data
X = pd.read_csv('irisx.csv', header=None)
y = pd.read_csv('irisy.csv', header=None)

# Use only first two features
X = X.iloc[:, 0:2]

# Split into train and test sets
X_train = X.iloc[:100, :]
X_test = X.iloc[100:, :]
y_train = y.iloc[:100]
y_test = y.iloc[100:]

# Function to train and evaluate perceptron with different penalties
def train_perceptron(penalty):
    # Initialize Perceptron with specified penalty
    clf = Perceptron(penalty=penalty, 
                     alpha=0.0001,  # Regularization parameter
                     max_iter=1000,
                     random_state=42)
    
    # Train the model
    clf.fit(X_train, y_train.values.ravel())
    
    # Make predictions
    y_pred = clf.predict(X_test)
    
    # Calculate accuracy
    accuracy = accuracy_score(y_test, y_pred)
    return round(accuracy, 2)

# Train and evaluate with L1 penalty
l1_accuracy = train_perceptron('l1')
print(f"Accuracy with L1 penalty: {l1_accuracy}")

# Train and evaluate with L2 penalty
l2_accuracy = train_perceptron('l2')
print(f"Accuracy with L2 penalty: {l2_accuracy}")
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import Perceptron
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score


X = pd.read_csv('irisX.csv', header=None).values
y = pd.read_csv('irisY.csv', header=None).values.ravel()  # Flatten the array

# Use only the first two features
X = X[:, :2]

# Split into training and testing sets
X_train, X_test = X[:100], X[100:]
y_train, y_test = y[:100], y[100:]

# Standardize the features
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Train and evaluate Perceptron with L1 and L2 penalty terms
results = {}
for penalty in ["l1", "l2"]:
    clf = Perceptron(penalty=penalty, max_iter=1000, tol=1e-3, random_state=42)
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    accuracy = round(accuracy_score(y_test, y_pred), 2)
    results[penalty] = accuracy

print(results)
```

```python
import pandas as pd
import numpy as np
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# # Load the dataset
# irisX_path = "path/to/irisX.csv"
# irisY_path = "path/to/irisY.csv"

X = pd.read_csv('irisX.csv', header=None).values[:, :3]  # Use first 3 features
y = pd.read_csv('irisY.csv', header=None).values.ravel()

# Split into training and testing sets (No feature normalization as per instructions)
X_train, X_test = X[:100], X[100:]
y_train, y_test = y[:100], y[100:]

# Train and evaluate SVM with RBF kernel, gamma=0.5, and different C values
C_values = [0.01, 1, 10]
best_accuracy = 0
best_C = None

for C in C_values:
    clf = SVC(kernel="rbf", gamma=0.5, C=C, decision_function_shape="ovr")
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    accuracy = round(accuracy_score(y_test, y_pred), 2)
    
    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_C = C

print(f"Best C: {best_C}, Best Accuracy: {best_accuracy}")
```

```python
import numpy as np
from sklearn.svm import SVC

# Define dataset (features and labels)
X_data = np.array([[1], [2], [4], [5], [6], [7], [9], [10]])  # Feature values (x)
y_data = np.array([1, 1, -1, -1, -1, -1, 1, 1])  # Output labels (y)

# Train SVC model with polynomial kernel (degree=3, C=1, gamma=0.1)
svm_model = SVC(kernel="poly", degree=3, C=1, gamma=0.1)
svm_model.fit(X_data, y_data)

# Get indices of support vectors
support_vector_indices = svm_model.support_

# Identify which points are NOT support vectors
non_support_vectors = [X_data[i][0] for i in range(len(X_data)) if i not in support_vector_indices]

# Print results
print("Support Vector Indices (0-based):", support_vector_indices)
print("Support Vectors (x-values):", X_data[support_vector_indices].flatten())
print("Non-Support Vectors (x-values):", non_support_vectors)
```

```python
import numpy as np
from sklearn.svm import SVC

# Define the dataset
X = np.array([[1], [2], [4], [5], [6], [7], [9], [10]])  # Feature values (x)
y = np.array([1, 1, -1, -1, -1, -1, 1, 1])              # Labels (y)

# Train the SVM model with the specified parameters
clf = SVC(kernel='poly', degree=3, C=1, gamma=0.1)
clf.fit(X, y)

# Get the indices of support vectors
support_vectors_indices = clf.support_
support_vectors = X[support_vectors_indices]

# Print support vectors
print("Support Vectors:", support_vectors)

# Check which points are not support vectors
non_support_vectors = [point for point in X if point not in support_vectors]
print("Non-Support Vectors:", non_support_vectors)
```
